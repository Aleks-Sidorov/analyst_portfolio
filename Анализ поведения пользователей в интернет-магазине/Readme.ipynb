## О проекте

Анализ поведения пользователей в интернет-магазине электроники.

**Цель**: понять, на каком этапе воронки магазин теряет пользователей, оценить удержание по когортам.

### Инструменты

**Python**

- `Pandas`
- `Matplotlib`
- `seaborn`
