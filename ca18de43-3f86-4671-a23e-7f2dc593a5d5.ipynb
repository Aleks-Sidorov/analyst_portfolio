## 2. Проверка гипотезы в Python

Гипотеза звучит так: пользователи из Санкт-Петербурга проводят в среднем больше времени за чтением и прослушиванием книг в приложении, чем пользователи из Москвы. Попробуйте статистически это доказать, используя одностороннюю проверку гипотезы с двумя выборками:

- Нулевая гипотеза H₀: Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.

- Альтернативная гипотеза H₁: Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.

```python
alpha = 0.05

stat_ttest, p_value_ttest = ttest_ind(
    df[df.city == 'Санкт-Петербург']['hours'],
    df[df.city == 'Москва']['hours'],
    alternative='greater'
)

if p_value_ttest > alpha:
    print(f'p-value t-теста = {round(p_value_ttest, 2)}')
    print('Нулевую гипотезу не удалось отвергнуть. Средняя активность пользователей в Москве и Санкт-Петербурге статистически значимо не различается.')
else:
    print(f'p-value t-теста ={round(p_value_ttest, 2)}')
    print('Нулевая гипотеза не находит подтверждения! Средняя активность пользователей в Санкт-Петербурге статистически значимо выше, чем в Москве.')
```

<div class="alert alert-warning"> <b>⚠️ Комментарий ревьюера v1:</b> <br>Одностороннее направление теста выбрано верно. Обрати внимание, что ttest_ind по умолчанию использует предположение о равенстве дисперсий. В аналитической записке стоит назвать выбранный вариант t-теста и обосновать его. Кроме того, при p-value выше уровня значимости корректнее говорить, что нулевую гипотезу не удалось отвергнуть, а не что она «нашла подтверждение». </div>

- достоверность теста — 95%.

```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

alpha = 0.05  # Уровень значимости
power = 0.8  # Мощность теста
p = 0.3 # Базовый уровень доли
mde = 0.03  # Минимальный детектируемый эффект
effect_size = proportion_effectsize(p, p + mde)

# Инициализируем класс NormalIndPower
power_analysis = NormalIndPower()

# Расчёт размера выборки
sample_size = power_analysis.solve_power(
    effect_size = effect_size,
    power = power,
    alpha = alpha,
    ratio = 1 # Равномерное распределение выборок
)

print(f"Необходимый размер выборки для каждой группы: {int(sample_size)}")
```

<div class="alert alert-warning"> <b>⚠️ Комментарий ревьюера v1:</b> <br>Необходимый размер выборки рассчитан корректно с учётом базовой конверсии, мощности и уровня значимости. Добавь сравнение полученного значения с фактическими размерами очищенных групп и сформулируй явный вывод о достаточности выборки. </div>

```python
group_A_size = participants_interface[participants_interface.group == 'A']['user_id'].nunique()
group_B_size = participants_interface[participants_interface.group == 'B']['user_id'].nunique()

print(f'Размер группы A: {group_A_size}')
print(f'Размер группы B: {group_B_size}')
print(f'Необходимый размер каждой группы: {int(sample_size)}')
```
